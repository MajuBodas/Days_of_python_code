```python
A = [1, 2, 3, 4, 5, 6]
```

```python
A = [-1, -3, -4, 0]
```

```python
#Solução:
def solution(A):
    biggest = A[0]
    
    for i in range(len(A)):
        if A[i] > biggest:
            biggest = A[i]
            
    if biggest > 0:  
        final_number = biggest + 1
        print(biggest)
    if biggest <= 0:
        final_number = biggest - biggest + 1 
        print(biggest)
      
    return final_number

print('final_number: ', solution(A))
```
